```python
#so that didn't seem to work. I think it might be the patch size that we are reading in
from deepforest import preprocess
import tempfile
import glob
import pandas
import os
tmpdir = tempfile.gettempdir()
RGBs = glob.glob("deepforest/RGB_training/*.tiff")
anns = pandas.read_csv("deepforest/annotations_file3.csv")
annotations_list = []
for i in RGBs :
  name = os.path.basename(i)
  anns_temp = anns.loc[anns['image_path'].eq(name)]
  tmp = preprocess.split_raster(annotations_file=anns_temp,
                                path_to_raster = i,
                                            save_dir=tmpdir,
                                            patch_size=400,
                                            patch_overlap=0.05,
                                            allow_empty=False,
                                root_dir = "deepforest/RGB_training/")
  # Returns a 6 column pandas array
  #print(tmp)
  tmp = tmp.iloc[:, :-1]
  #print(tmp)
  annotations_list.append(tmp)
```

```python
# Concatenate the list of dataframes into a single dataframe
all_annotations = annotations_list[0]
for i in range(1, len(annotations_list)):
  all_annotations = pandas.concat([all_annotations, annotations_list[i]])

# Display the first few rows of the concatenated dataframe
display(all_annotations.head())
all_annotations.to_csv("deepforest/all_annotations.csv")
print(all_annotations)
```
